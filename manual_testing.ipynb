```python
import numpy as np
values = np.array([943.0518, 943.0518, 942.0518, 1042.0518, 1117.0518, 1172.0518, 1196.0518, 1215.0518])
```

```python
x = np.arange(len(values))
x
```

```python
# fit a straight line through the points
slope, intercept = np.polyfit(x, values, 1)
print(f"Slope     : {slope:.4f}")
print(f"Intercept : {intercept:.4f}")

# predicted values on the fitted line
y_pred = slope * x + intercept
print(f"Predicted : {y_pred.round(2)}")

# SS_residual — how far actual values deviate from the line
ss_res = np.sum((values - y_pred) ** 2)
print(f"SS_residual: {ss_res:.4f}")

# SS_total — total variation in the data around the mean
ss_tot = np.sum((values - values.mean()) ** 2)
print(f"SS_total   : {ss_tot:.4f}")

# R-squared
r_squared = 1 - (ss_res / ss_tot)
print(f"R²         : {r_squared:.4f}")
```
